## Project Overview

Lending plays an important role in economic growth by providing capital for investments, consumer spending, and infrastructure. However, lending also carries the risk that borrowers may not repay their loans. Effective credit risk management helps lenders support growth while maintaining financial stability, reputation, and customer trust.

### Objective

The goal of this project is to build a web application where users can enter loan application details and receive:

- The predicted probability of loan approval  
- An explanation of how each input feature influenced the prediction  

### Model Interpretability

To make model decisions transparent and trustworthy, this project uses interpretability techniques such as:

- **SHAP** (SHapley Additive exPlanations)  
- **LIME** (Local Interpretable Model-agnostic Explanations)  

Model interpretability is essential because:

- **For lenders:** It provides confidence in automated credit decisions by explaining the reasoning behind predictions.  
- **For borrowers:** It offers clear insights into factors affecting approval chances, helping them improve future applications.

## Dataset Information

The dataset used in this project is the **Loan Approval Classification Dataset** ([Kaggle link](https://www.kaggle.com/datasets/taweilo/loan-approval-classification-data)).

It is a synthetic dataset inspired by:

- **Credit Risk Dataset**  
  https://www.kaggle.com/datasets/laotse/credit-risk-dataset  
- **Financial Risk for Loan Approval Dataset**  
  https://www.kaggle.com/datasets/lorenzozoppelletto/financial-risk-for-loan-approval  

### Key Characteristics

- Contains both **categorical** and **numerical** features  
- Provides a diverse set of variables relevant to loan approval  
- Synthetic data makes it suitable for learning and experimentation, since real credit data is often restricted due to privacy and regulatory constraints

```python
# Import the kagglehub library, which provides utilities to access Kaggle datasets programmatically
import kagglehub

# Import the KaggleDatasetAdapter class specifically for adapting/handling Kaggle datasets
from kagglehub import KaggleDatasetAdapter

# Import Path from pathlib for object-oriented filesystem path handling
from pathlib import Path

# Set root_dir to the parent directory of the current working directory.
root_dir = Path.cwd().parent
```

### Dataset Download Approach

The dataset is downloaded directly from Kaggle using the Kaggle API and loaded into a pandas DataFrame with the `KaggleDatasetAdapter`.

**Workflow**
- Access dataset via Kaggle API
- Download data programmatically
- Load into a pandas DataFrame for analysis

```python
# Load a dataset file from Kaggle directly into a Python object using kagglehub.
df = kagglehub.dataset_load(
    adapter = KaggleDatasetAdapter.PANDAS,                 # Use the Pandas adapter → returns a pandas DataFrame
    handle = "taweilo/loan-approval-classification-data",  # Kaggle dataset identifier: <owner>/<dataset-name>
    path = "loan_data.csv"                                 # Specific file within the dataset to load
)

# Quick data inspection
df.head()
```

### Column Information

The dataset contains **45,000 records** and **14 columns**, described below.

| Column | Description | Type |
|--------|------------|------|
| `person_age` | Age of the person | Float |
| `person_gender` | Gender of the person | Categorical |
| `person_education` | Highest education level | Categorical |
| `person_income` | Annual income | Float |
| `person_emp_exp` | Years of employment experience | Integer |
| `person_home_ownership` | Home ownership status (rent, own, mortgage, etc.) | Categorical |
| `loan_amnt` | Loan amount requested | Float |
| `loan_intent` | Purpose of the loan | Categorical |
| `loan_int_rate` | Loan interest rate | Float |
| `loan_percent_income` | Loan amount as a percentage of annual income | Float |
| `cb_person_cred_hist_length` | Length of credit history (years) | Float |
| `credit_score` | Credit score | Integer |
| `previous_loan_defaults_on_file` | Indicator of previous loan defaults | Categorical |
| `loan_status` *(target)* | Loan approval status: **1 = approved, 0 = rejected** | Integer |


The dataset is saved locally to ensure consistent access and reuse across subsequent notebooks in this project.

```python
# Construct the output file path where the raw dataset CSV will be saved.
file_path = root_dir / "datasets" / "raw_data.csv"

# Ensure that the parent directory ("datasets") exists before writing the file.
# parents=True → create any missing intermediate directories
# exist_ok=True → do not raise an error if the directory already exists
file_path.parent.mkdir(parents=True, exist_ok=True)

# Persist the DataFrame to disk as a CSV file.
# index=False → do not write the pandas index column to the CSV
df.to_csv(file_path, index=False)

print(f"Dataset saved to {file_path}")
```
